```python
from fuzzywuzzy import fuzz
```

```python
text1 = 'The quick brown fox jumps over the lazy dog'
text2 = 'The quick brown fox jumps over the lazy dog.'

similarity = fuzz.ratio(text1,text2)
print(similarity)
```

```python
#mencari potongan substring
similarity = fuzz.partial_ratio(text1,text2)
print(similarity)
```

```python
str1 = "My name is ali"
str2 = "Ali is my name"

ratio = fuzz.ratio(str1.lower(),str2.lower())
print(ratio)
```

```python
str1 = "My name is ali"
str2 = "My name is ali aaaa"
print(fuzz.partial_ratio(str1.lower(),str2.lower()))
```

```python
str1 = "My name is ali"
str2 = "Ali is my name"

print(fuzz.token_sort_ratio(str1,str2))
```

```python
str1 = "My name is ali"
str2 = "Ali is my name name"

print(fuzz.token_sort_ratio(str1,str2))
print(fuzz.token_set_ratio(str1,str2))
```

```python
str_a = 'Read the sentence - My name is Ali'
str_b = 'My name is Ali'

ratio = fuzz.token_set_ratio(str_a,str_b)
print(ratio)
```

```python
from fuzzywuzzy import process
query = 'My name is Ali'
choices = ['My name Ali','My name is Ali','My ali']

process.extract(query,choices)
```
