```python
 import matplotlib.pyplot as plt

 import seaborn as sns

 import numpy as np

 from sklearn.datasets import make_blobs

 from sklearn.cluster import KMeans
```

```python
X, y_true = make_blobs(n_samples=300, n_features=2, centers=4, cluster_std=0.60, random_state=0)

plt.scatter(X[:, 0], X[:, 1], s=50)
```

```python
kmeans = KMeans(n_clusters=5)
kmeans.fit(X)
```

```python
y_kmeans = kmeans.predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], s=50, c=y_kmeans)
```

```python
mapping = {0: "A", 1: "B", 2: "C", 3: "D", 4: "E"}
y_labels = [mapping[label] for label in y_kmeans]
```

```python
plt.scatter(X[:,0], X[:,1], s=50, c=y_kmeans, cmap='viridis')
```
